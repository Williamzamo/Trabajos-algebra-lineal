**Taller 7 programacion**

```python
#Taller 07
class Accion:
    def __init__(self, tipo, descripcion):
        self.tipo = tipo
        self.descripcion = descripcion

    def __str__(self):
        return f"[{self.tipo}] {self.descripcion}"

class Nodo:
    def __init__(self, accion):
        self.accion = accion
        self.anterior = None

class Pila:
    def __init__(self):
        self.tope = None

    def esta_vacia(self):
        return self.tope is None

    def agregar(self, accion):
        nuevo_nodo = Nodo(accion)
        nuevo_nodo.anterior = self.tope
        self.tope = nuevo_nodo

    def recuperar(self):
        if self.esta_vacia():
            print("La pila está vacía")
            return None
        accion = self.tope.accion
        self.tope = self.tope.anterior
        return accion

    def ver_accion(self):
        if self.esta_vacia():
            print("La pila está vacía")
            return None
        return self.tope.accion

    def mostrar(self):
        if self.esta_vacia():
            print("La pila está vacía.")
            return
        nodo_actual = self.tope
        while nodo_actual is not None:
            etiqueta = "tope" if nodo_actual is self.tope else ""
            print(f"{nodo_actual.accion}{etiqueta}")
            nodo_actual = nodo_actual.anterior

def mostrar_menu():
    print("1.Registrar nueva acción")
    print("2.Deshacer última acción")
    print("3.Rehacer acción deshecha")
    print("4.Ver última acción")
    print("5.Mostrar historial completo")
    print("6.Salir")
    print("Elige una opción: ", end="")


def pedir_accion():
    """Le pide al usuario el tipo y descripción de la acción."""
    tipo = input("Tipo de acción: ")
    if not tipo:
        tipo = "genérica"
    descripcion = input("Descripción: ")
    if not descripcion:
        descripcion = "Sin descripción"
    return Accion(tipo, descripcion)

pila_1 = Pila()
pila_historial = Pila()

while True:
    mostrar_menu()
    opcion = input()

    if opcion == "1":
        print("registrar accion")
        accion = pedir_accion()
        pila_1.agregar(accion)
        pila_historial = Pila()
        print(f"Acción registrada: {accion}")

    elif opcion == "2":
        print("deshacer accion")
        accion = pila_1.recuperar()
        if accion:
            pila_historial.agregar(accion)
            print(f"Acción deshecha: {accion}")

    elif opcion == "3":
        print("reacer accion")
        accion = pila_historial.recuperar()
        if accion:
            pila_1.agregar(accion)
            print(f"Acción rehecha: {accion}")

    elif opcion == "4":
        print("ultima accion")
        accion = pila_1.ver_accion()
        if accion:
            print(f"{accion}")

    elif opcion == "5":
        print("historial, (de reciente a antigua)")
        pila_1.mostrar()

    elif opcion == "6":
        break

    else:
        print("Opción inválida. Elige un número del 1 al 6.")
```
