```python

import math
import random
import matplotlib.pyplot as plt

L1 = [(-0.14, -0.17), (-0.11, 0.7), (-0.13, -1.5), (0.33, -0.27), (-0.22, 0.12), (0.23, 1.16), (0.66, 0.11), (-0.74, -1.01), (0.25, 1.31), (0.04, -0.11), (0.53, -1.45), (-0.31, 0.49), (0.87, -0.24), (0.38, 0.25), (0.78, -1.11), (0.57, -1.51), (-2.62, -0.61), (-0.92, 0.88), (0.66, -1.22), (0.85, -1.0), (-0.09, -0.29), (0.11, 0.82), (0.64, 0.35), (0.65, 0.48), (-0.63, -0.72), (-0.47, 0.5), (-0.25, 2.34), (-0.82, -1.1), (0.77, 1.42), (0.51, 0.84), (1.43, -0.09), (-1.42, -0.53), (0.95, -1.44), (0.03, 0.25), (-0.32, 0.72), (0.58, 2.32), (0.62, -0.61), (-0.56, -0.83), (0.95, -0.57), (-0.07, 0.75), (9.28, 9.71), (8.16, 8.92), (9.43, 10.42), (11.19, 9.98), (10.26, 10.17), (11.08, 10.89), (10.27, 8.99), (10.9, 10.38), (11.23, 9.97), (11.95, 9.64), (11.59, 10.12), (9.48, 8.87), (9.85, 11.42), (10.82, 10.69), (7.62, 10.71), (10.56, 9.45), (9.37, 10.0), (11.72, 8.94), (9.57, 11.36), (9.55, 9.64), (10.1, 8.76), (10.22, 8.79), (10.89, 10.0), (12.28, 10.28), (11.37, 8.7), (9.88, 10.32), (11.75, 8.32), (10.99, 10.59), (11.53, 10.71), (10.05, 9.48), (8.75, 10.2), (9.81, 12.02), (9.39, 10.32), (8.43, 9.6), (10.26, 10.82), (11.45, 9.96), (8.88, 10.46), (10.52, 10.49), (9.3, 11.13), (10.09, 10.7), (11.27, 0.61), (10.29, 2.15), (10.24, -0.3), (10.11, 1.48), (10.12, 0.52), (11.2, -0.51), (8.27, 0.3), (10.23, -0.61), (10.87, 0.61), (9.01, 0.52), (9.8, -1.48), (10.45, -0.05), (9.25, 0.52), (10.49, -0.58), (10.4, 1.02), (9.29, 0.38), (10.0, 2.27), (8.14, 0.7), (9.69, -0.1), (11.9, -0.01), (12.24, -0.44), (10.35, -0.49), (9.31, 0.06), (9.72, 0.2), (7.89, 2.0), (9.83, 1.74), (8.99, 0.29), (13.16, -0.87), (8.5, -0.55), (10.47, 0.69), (11.29, -0.26), (8.4, -0.44), (11.24, 0.47), (8.04, -0.04), (11.4, 2.13), (10.56, 0.31), (8.76, -0.84), (10.05, 0.5), (10.58, -0.48), (8.85, -0.78)]
L2 = [(4.19, 5.84), (5.23, 9.35), (2.04, 7.16), (2.39, 3.96), (6.72, 3.0), (3.16, 7.52), (0.73, 4.58), (9.98, 9.96), (0.73, 2.13), (2.65, 9.33), (8.81, 8.79), (3.7, 1.58), (8.34, 7.04), (6.12, 9.87), (6.54, 0.08), (8.17, 2.99), (6.63, 9.39), (1.34, 1.15), (1.07, 5.53), (2.72, 6.05), (7.18, 2.04), (6.34, 2.64), (4.89, 9.05), (8.46, 0.92), (4.24, 2.77), (0.04, 7.71), (6.37, 2.62), (7.41, 5.52), (4.28, 0.1), (0.75, 8.83), (9.04, 5.46), (8.35, 5.83), (1.48, 1.27), (3.08, 8.99), (7.96, 8.61), (8.99, 2.1), (2.5, 1.03), (7.8, 8.84), (4.06, 6.21), (1.55, 9.3), (8.65, 9.76), (8.11, 8.81), (0.25, 7.37), (3.32, 9.31), (8.02, 8.64), (8.11, 2.67), (7.87, 1.08), (8.72, 8.59), (2.22, 8.17), (4.6, 3.05), (7.95, 2.28), (0.24, 1.93), (3.28, 8.64), (9.67, 2.79), (6.41, 4.0), (9.81, 5.36), (9.39, 1.15), (9.7, 1.79), (9.63, 2.65), (1.08, 4.35), (7.29, 3.14), (6.06, 5.11), (3.85, 5.77), (2.55, 7.09), (0.02, 9.26), (5.38, 7.19), (7.42, 6.71), (3.64, 0.7), (6.64, 3.3), (3.14, 8.48), (7.2, 3.0), (3.09, 4.08), (4.02, 2.96), (1.27, 4.2), (9.4, 6.77), (9.03, 6.16), (3.01, 5.48), (0.0, 2.87), (4.3, 5.8), (6.55, 4.65), (4.42, 2.14), (4.73, 9.01), (7.96, 1.7), (0.85, 5.15), (6.33, 3.35), (8.18, 7.51), (6.73, 2.25), (1.99, 0.24), (2.45, 4.75), (8.5, 0.73), (4.14, 6.3), (1.94, 6.96), (4.94, 2.44), (6.56, 0.06), (7.51, 7.7), (1.07, 4.25), (1.76, 9.58), (5.18, 0.5), (2.49, 8.48), (4.56, 8.01), (6.68, 9.88), (5.95, 9.5), (8.91, 6.13), (7.19, 5.05), (8.31, 5.48), (8.97, 7.44), (4.75, 2.59), (2.47, 6.38), (7.66, 5.21), (6.27, 2.75), (0.77, 2.86), (2.72, 3.2), (5.4, 1.38), (2.31, 6.94), (7.06, 0.64), (4.08, 5.43), (4.16, 2.07), (4.2, 9.05), (5.84, 6.96), (8.57, 7.66), (3.8, 0.06), (3.52, 7.53), (8.53, 9.53), (4.19, 7.48), (5.46, 6.03), (2.21, 2.19), (4.36, 0.29), (3.36, 6.79), (4.04, 1.65), (4.67, 1.28), (6.22, 0.27), (3.94, 5.64), (0.27, 6.43), (1.36, 4.62), (0.5, 3.79), (2.12, 3.27), (7.61, 3.79), (7.52, 8.32), (2.52, 0.82), (0.19, 5.39), (10.0, 3.5), (6.5, 7.81), (6.52, 7.54), (9.5, 1.99), (0.2, 1.52), (1.26, 6.69), (5.64, 2.18), (6.99, 7.67), (1.68, 6.07), (7.48, 1.15)]
L3 = [(5.87, 3.12), (6.96, 5.32), (4.3, 6.7), (3.68, 6.0), (4.52, 2.85), (3.53, 3.49), (6.64, 6.22), (3.62, 3.99), (6.64, 6.56), (3.87, 6.62), (3.31, 5.82), (3.85, 6.79), (6.19, 3.35), (6.88, 4.53), (3.61, 3.41), (3.35, 5.71), (4.78, 2.89), (5.71, 3.16), (3.21, 4.56), (4.76, 3.0), (4.75, 7.15), (6.52, 5.92), (6.48, 3.64), (3.33, 3.61), (3.34, 6.58), (5.85, 6.87), (5.48, 6.98), (6.37, 3.72), (4.4, 3.1), (6.91, 4.35), (6.69, 6.01), (5.09, 3.17), (4.31, 3.24), (5.5, 3.21), (6.7, 6.13), (4.23, 7.09), (3.42, 4.11), (3.19, 5.92), (4.95, 3.11), (4.71, 7.19), (4.98, 7.07), (6.51, 6.41), (3.04, 4.55), (5.4, 6.86), (3.07, 5.19), (3.42, 4.75), (4.57, 7.03), (5.29, 7.04), (5.94, 6.97), (4.69, 6.99), (6.76, 4.71), (6.35, 6.37), (6.15, 3.75), (3.2, 4.09), (3.12, 5.37), (6.93, 5.65), (6.61, 4.38), (3.35, 6.19), (6.76, 5.88), (6.16, 6.58), (3.3, 4.29), (6.8, 6.66), (5.16, 3.22), (5.28, 6.81), (3.17, 4.74), (6.31, 3.42), (6.92, 4.88), (3.3, 3.6), (4.87, 2.91), (5.48, 6.91), (3.91, 3.27), (6.85, 5.94), (6.81, 5.03), (4.47, 6.99), (5.22, 6.92), (3.3, 5.51), (4.14, 2.94), (3.67, 3.67), (3.86, 3.18), (3.14, 4.46), (1.86, 0.81), (9.37, 7.1), (7.48, 9.24), (0.25, 2.8), (3.76, 10.01), (3.4, 0.09), (9.52, 3.3), (9.37, 7.42), (9.73, 6.21), (4.99, 10.05), (9.43, 7.86), (0.61, 2.76), (9.98, 3.41), (4.39, 9.96), (9.85, 6.25), (2.2, 1.06), (9.39, 2.44), (1.57, 1.12), (-0.11, 5.27), (2.46, 9.4), (2.55, 0.54), (3.3, 0.41), (3.75, 10.11), (0.33, 3.48), (0.25, 6.39), (10.03, 4.05), (1.25, 7.58), (2.28, 0.85), (0.26, 3.7), (1.74, 8.85), (5.2, 0.18), (6.39, 9.62), (0.1, 3.43), (3.43, 0.33), (8.76, 8.6), (5.33, 9.95), (7.62, 9.04), (9.12, 7.73), (10.14, 4.75), (8.63, 8.31), (7.35, 0.86), (0.01, 4.72), (4.24, 10.16), (5.18, 9.63), (0.5, 3.5), (-0.22, 4.73), (8.69, 1.51), (6.22, 0.25), (0.56, 7.63), (7.11, 0.55), (3.99, 9.84), (10.19, 4.96), (-0.1, 5.33), (2.1, 9.26), (0.41, 3.45), (2.84, 9.46), (6.86, 9.57), (5.5, 9.79), (5.84, -0.17), (3.59, 0.13), (6.82, 0.52), (7.21, 0.35), (9.92, 4.4), (-0.0, 4.93), (0.61, 2.2), (0.56, 6.75), (0.29, 4.23), (0.72, 7.71), (2.13, 0.79), (10.04, 4.01), (2.4, 9.18), (7.86, 1.15), (8.05, 0.92), (3.01, 9.46), (4.02, 0.08), (9.58, 6.05), (9.34, 6.99), (9.75, 4.92), (4.9, 0.06), (1.65, 1.26)]

def distancia_cuadrada(p, q):
    return sum((pi - qi) ** 2 for pi, qi in zip(p, q))


def centroide(grupo):
    n = len(grupo[0])  # dimensión de los puntos
    suma = [0.0] * n
    for punto in grupo:
        for j in range(n):
            suma[j] += punto[j]
    return tuple(coord / len(grupo) for coord in suma)


def calcular_C(grupos, representantes):
    N = sum(len(g) for g in grupos)
    if N == 0:
        return 0.0
    suma = 0.0
    for grupo, r in zip(grupos, representantes):
        for x in grupo:
            suma += distancia_cuadrada(x, r)
    return suma / N


#ALGORITMO DE LAS K-MEDIAS
def kmeans(L, k, max_iter=100):
    # --- Paso 1: representantes iniciales, elegidos al azar entre los puntos ---
    representantes = random.sample(L, k)

    # Guardamos los grupos de la iteración anterior para detectar convergencia
    grupos_anteriores = None
    historial_C = []

    for iteracion in range(max_iter):
        # --- Paso 2: asignar cada punto al representante más cercano ---
        grupos = [[] for _ in range(k)]
        for x in L:
            mejor_indice = 0
            mejor_distancia = distancia_cuadrada(x, representantes[0])
            for i in range(1, k):
                d = distancia_cuadrada(x, representantes[i])
                if d < mejor_distancia:  # estrictamente menor -> en empate
                    mejor_distancia = d  # se queda el de menor índice
                    mejor_indice = i
            grupos[mejor_indice].append(x)

        historial_C.append(calcular_C(grupos, representantes))

        # ¿Convergió? (los grupos son idénticos a los de la iteración anterior)
        if grupos == grupos_anteriores:
            break

        # --- Paso 3: recalcular representantes como centroides ---
        nuevos_representantes = []
        for i in range(k):
            if len(grupos[i]) == 0:
                # Grupo vacío: no se puede calcular centroide, se conserva
                nuevos_representantes.append(representantes[i])
            else:
                nuevos_representantes.append(centroide(grupos[i]))

        representantes = nuevos_representantes
        grupos_anteriores = grupos

    return grupos, representantes, historial_C



#REPETIR K-MEDIAS m VECES
def kmeans_repetido(L, k, m=10, max_iter=100):
    mejor_grupos = None
    mejor_representantes = None
    mejor_C = math.inf
    valores_C_intentos = []

    for _ in range(m):
        grupos, representantes, _ = kmeans(L, k, max_iter=max_iter)
        C = calcular_C(grupos, representantes)
        valores_C_intentos.append(C)

        if C < mejor_C:
            mejor_C = C
            mejor_grupos = grupos
            mejor_representantes = representantes

    return mejor_grupos, mejor_representantes, mejor_C, valores_C_intentos



#GRAFICAR RESULTADOS
def graficar_resultado(grupos, representantes, titulo="Resultado de k-medias"):
    colores = plt.cm.tab10.colors

    plt.figure(figsize=(6, 6))
    for i, grupo in enumerate(grupos):
        if len(grupo) == 0:
            continue
        xs = [p[0] for p in grupo]
        ys = [p[1] for p in grupo]
        color = colores[i % len(colores)]
        plt.scatter(xs, ys, color=color, s=25, label=f"Grupo {i + 1}")
    rx = [r[0] for r in representantes]
    ry = [r[1] for r in representantes]
    plt.scatter(rx, ry, color="black", marker="X", s=150,label="Representantes", zorder=5)


    plt.title(titulo)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.axis("equal")
    plt.tight_layout()


L = L1
k = 1
grupos, representantes, C, valores_C = kmeans_repetido(L, k, m=10)
graficar_resultado(grupos, representantes)
print(valores_C)
print(C)
plt.show()
```
